Load results for the Disco experiment

```python
import os
import numpy as np
```

```python
dataset="Disco/no_tracto/"
TRACT='q'

e1='7e-4'
e2='2e-5'
e3='3e-4'


W1=['0','0.01','0.03','0.07'] #colinear to FW (II)
W2=['0','0.002'] #colinear to iso (IV)
W3=['0','0.05','0.1'] #colinear to reduced (V)
LOW1=['1','3'] #reduced signal (III)
LOW2=['1','3','6'] #colinear to reduced (V)
T=['0','0.05','0.1'] #Isotropification


names=['cores_FA_nolift','cores_wColinear_nolift','cores_wFW_nolift',
       'cores_MD_nolift','cores_AD_nolift','cores_RD_nolift','cores_ADC_nolift',
       'cores_MAS_nolift']

L=[1,1,1,1,1,
   1,1,
   1]

K=[1,1,1,1,1,
   1,1,
   None]

beta=1
alpha=0.0
sum_or_max='sum'
```

```python
n_seed=1000
n_neighbors=3
train_size=0.75

path="/home/gui/Documents/Post-doc/data/"+dataset

res=np.zeros((len(names),len(W1),len(W2),len(W3),len(LOW1),len(LOW2),len(T)))
for n,name in enumerate(names):
    for i1,w1 in enumerate(W1):
        for i2,w2 in enumerate(W2):
            for i3,w3 in enumerate(W3):
                for j1,l1 in enumerate(LOW1):
                    for j2,l2 in enumerate(LOW2):
                        for k,t in enumerate(T):
                            path_temp=path+'e'+e1+'_e'+e2+'_e'+e3+'/w'+w1+'_'+w2+'_'+w3+'_l'+l1+'_'+l2+'_t'+t+'/matrices_distance/res/'
                            d1='e'+e1+'_e'+e2+'_e'+e3+'_healthy'
                            d2='e'+e1+'_e'+e2+'_e'+e3+'_w'+w1+'_'+w2+'_'+w3+'_l'+l1+'_'+l2+'_t'+t+'_ill'

                            if os.path.isfile(path_temp+d1+d2+'_'+name+'_L'+str(L[n])+'_k'+str(K[n])+'_beta'+str(beta)+'_'+sum_or_max+'_s'+str(n_seed)+'_n'+str(n_neighbors)+'_t'+str(train_size)+'.txt'):
                                res[n,i1,i2,i3,j1,j2,k]=np.loadtxt(path_temp+d1+d2+'_'+name+'_L'+str(L[n])+'_k'+str(K[n])+'_beta'+str(beta)+'_'+sum_or_max+'_s'+str(n_seed)+'_n'+str(n_neighbors)+'_t'+str(train_size)+'.txt')
print(res.shape)
```

```python
#========== FA =============
#-1 et 0 parceque ==1 -2 parce pas fais
# PAs fais dans les -2 w1
# Conclusion MAS mieux sauf pour T 
print((res[-1,:-2,:,:,:,:,:]-res[0,:-2,:,:,:,:,:]).sum())
print(res[-1,0,0,1,0,1,0],res[0,0,0,1,0,1,0])
```

```python
#========== wColinear =============
# 0 parce que poids 
# En particulier bien pour LOW1

print((res[-1,0,0,0,:,:,:]-res[1,0,0,0,:,:,:]).sum())
print(res[-1,0,0,0,:,:,:]-res[1,0,0,0,:,:,:])
```

```python
#========== wFW =============

print((res[-1,0,:,:,:,:,:]-res[2,0,:,:,:,:,:]).sum())
print(res[-1,0,:,:,:,:,:]-res[2,0,:,:,:,:,:])
```

```python
#========== MD =============
# MD better for w1>0
print((res[-1,0,:,:,:,:,:]-res[3,0,:,:,:,:,:]).sum())
#print(res[-1,:,:,:,:,:,:]-res[3,:,:,:,:,:,:])
```
