Notebook to motivate the use of multiple cores augmented streamlines

```python
import numpy as np
from dipy.io.streamline import load_tractogram,save_tractogram
import dipy
```

```python
path='/home/gui/Documents/Post-doc/data/tiny_Disco/100'
tract='r'
```

```python
t=load_tractogram('/home/gui/Documents/Post-doc/data/Disco/100/05-Tracts/trk/'+tract+'.trk',
                  reference='same',bbox_valid_check=False)
```

```python
ta=dipy.io.stateful_tractogram.StatefulTractogram(streamlines=t.streamlines,reference=path+'/02-DTI/FA_100.nii.gz',space=t.space)
ta.space_attributes,ta.affine,ta.dimensions,ta.voxel_sizes,ta.voxel_order
```

```python
save_tractogram(ta,path+'/05-Tracts/trk/'+tract+'.trk',bbox_valid_check=False)
```
