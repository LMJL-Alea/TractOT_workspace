```python
import numpy as np
from dipy.io.streamline import load_tractogram,save_tractogram

from tractot.load_to_numpy import trx_to_numpy_SAT_array,trx_to_numpy_MAT_array
from tractot.build_cores import infer_labels_svm_endpoint,average_SAS_per_labels,average_MAS_per_labels
from tractot.utils import random_MASs
```

# cores MAS

```python
d=3
k=4
nb_MAS=100
nb_MAP=40
m,wI,I,w,S=random_MASs(nb_MAS,nb_MAP,k,d)
conf=np.ones((nb_MAS,nb_MAP))
```

```python
L=5
seed=0
k2=5
```

```python
labels_svm,labels_clus=infer_labels_svm_endpoint(None,L,m,random_state=seed)
print(labels_svm.shape,labels_clus.shape)
```

```python
cores_m,cores_wI,cores_I,cores_w,cores_S,idx_svm,idx_cluster,size_cluster=average_MAS_per_labels(L,m,wI,I,w,S,conf,labels_svm,labels_clus,k2=k2)
```

```python
print(len(cores_m))
print(cores_m[0].shape,cores_wI[0].shape,cores_I[0].shape,cores_w[0].shape,cores_S[0].shape,idx_svm[0].shape,idx_cluster[0].shape,size_cluster[0].shape)
```

```python
Scalar=np.random.rand(nb_MAS,nb_MAP)
cores_m,cores_w,cores_Scalar,idx_svm,idx_cluster,size_cluster,idx_antecedant=average_SAS_per_labels(L,m,Scalar,labels_svm,labels_clus,k=k2,antecedant=True)
```
