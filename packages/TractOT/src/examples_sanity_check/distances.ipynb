Example of distance between different objects (MAP, MAS, MAT,...)

```python
import numpy as np

from tractot.distances import distances_aniso,distances_aniso_along,distances_iso_aniso,distances_iso_aniso_along
from tractot.distances import distances_spatial_iso_aniso,distances_spatial_iso_aniso_along
from tractot.distances import distance_MAP
from tractot.distances import distance_MAS_idx,distances_MAS_idx,distances_MAS_svm,distance_coresMAS_svm
from tractot.distances import distances_FAS_svm,distance_coresFA_svm

from tractot.utils import random_MASs
```

# Centroid MASs

```python
d=3
k1,k2=4,5
nb_MAS1,nb_MAS2=10,11 #nb streamline
nb_MAP1,nb_MAP2=20,21 #nb points on streamline
m1,wI1,I1,w1,S1=random_MASs(nb_MAS1,nb_MAP1,k1,d)
m2,wI2,I2,w2,S2=random_MASs(nb_MAS2,nb_MAP2,k2,d)
print(m1.shape,wI1.shape,I1.shape,w1.shape,S1.shape)
print(m2.shape,wI2.shape,I2.shape,w2.shape,S2.shape)
```

```python
#matrix of distances between two covariates
print(distances_aniso(S1[0,0],S2[0,0]).shape)
print(distances_aniso(S1[0,0],S2[0,0])[0,0])

#matrices of distances between two covariates along a streamline
print(distances_aniso_along(S1[0],S2[0,:20]).shape)
print(distances_aniso_along(S1[0],S2[0,:20])[0,0,0])
```

```python
#matrix of distances between two covariates and isotropic compartment
print(distances_iso_aniso(I1[0,0],I2[0,0],S1[0,0],S2[0,0]))

#matrix of distances between two covariates and isotropic compartment along a streamline
print(distances_iso_aniso_along(I1[0],I2[0,:20],S1[0],S2[0,:20])[0])
print(distances_iso_aniso_along(I1[0],I2[0,:20],S1[0],S2[0,:20]).shape)
```

```python
#matrix of distances between two spatials, covariates and isotropic compartment
print(distances_spatial_iso_aniso(m1[0,0],m2[0,0],I1[0,0],I2[0,0],S1[0,0],S2[0,0]))
print(distances_spatial_iso_aniso(m1[0,0],m2[0,0],I1[0,0],I2[0,0],S1[0,0],S2[0,0]).shape)

#matrix of distances between two spatials, covariates and isotropic compartment along a streamline
print(distances_spatial_iso_aniso_along(m1[0],m2[0,:20],I1[0],I2[0,:20],S1[0],S2[0,:20])[0])
print(distances_spatial_iso_aniso_along(m1[0],m2[0,:20],I1[0],I2[0,:20],S1[0],S2[0,:20]).shape)
```

```python
# distance between 2 MAPs 
print(distance_MAP(m1[0,0],m2[0,0],wI1[0,0],wI2[0,0],I1[0,0],I2[0,0],w1[0,0],w2[0,0],S1[0,0],S2[0,0],alpha=.5))
```

```python
#Distance between 2 MAS (with aligned indices)
print(distance_MAS_idx(m1[0],m2[0,:20],wI1[0],wI2[0,:20],I1[0],I2[0,:20],w1[0],w2[0,:20],S1[0],S2[0,:20],alpha=.5))

#Matrix of distance between MAS
print(distances_MAS_idx(m1,m2[:,:20],wI1,wI2[:,:20],I1,I2[:,:20],w1,w2[:,:20],S1,S2[:,:20],alpha=0.5).shape)
```

```python
# Distance between cores MAS
d=3
k1,k2=4,5
L=10

cores1_m,cores1_wI,cores1_I,cores1_w,cores1_S,labels_svm1=[],[],[],[],[],[]
cores2_m,cores2_wI,cores2_I,cores2_w,cores2_S,labels_svm2=[],[],[],[],[],[]


for i in range(L):
    nb_pts1_i=np.random.randint(10,41)
    nb_pts2_i=np.random.randint(10,41)

    
    cores1_m_i,cores1_wI_i,cores1_I_i,cores1_w_i,cores1_S_i=random_MASs(1,nb_pts1_i,k1,d)
    cores2_m_i,cores2_wI_i,cores2_I_i,cores2_w_i,cores2_S_i=random_MASs(1,nb_pts2_i,k2,d)
    
    cores1_m+=[cores1_m_i[0]]
    cores1_wI+=[cores1_wI_i[0]]
    cores1_I+=[cores1_I_i[0]]
    cores1_w+=[cores1_w_i[0]]
    cores1_S+=[cores1_S_i[0]]
    labels_svm1+=[np.random.choice(np.arange(1,41), size=nb_pts1_i, replace=False)]
    
    cores2_m+=[cores2_m_i[0]]
    cores2_wI+=[cores2_wI_i[0]]
    cores2_I+=[cores2_I_i[0]]
    cores2_w+=[cores2_w_i[0]]
    cores2_S+=[cores2_S_i[0]]
    labels_svm2+=[np.random.choice(np.arange(1,41), size=nb_pts2_i, replace=False)]
```

```python
print(len(cores1_m))
i=5
print(cores1_m[i].shape,cores1_wI[i].shape,cores1_I[i].shape,cores1_w[i].shape,cores1_S[i].shape,labels_svm1[i].shape)
```

```python
C,C_nb_MAP=distances_MAS_svm(cores1_m,cores2_m,cores1_wI,cores2_wI,cores1_I,cores2_I,
                             cores1_w,cores2_w,cores1_S,cores2_S,labels_svm1,labels_svm2,alpha=0.5)
```

```python
cost,sigma,C,C_nb_pts=distance_coresMAS_svm(cores1_m,cores2_m,cores1_wI,cores2_wI,cores1_I,cores2_I,
                             cores1_w,cores2_w,cores1_S,cores2_S,labels_svm1,labels_svm2,alpha=1)
print(cost)
```

```python
# Distance between cores FAS
d=3
k1,k2=4,5
L=10

cores1_m,FA1,w1,labels_svm1=[],[],[],[]
cores2_m,FA2,w2,labels_svm2=[],[],[],[]


for i in range(L):
    nb_pts1_i=np.random.randint(30,41)
    nb_pts2_i=np.random.randint(30,41)

    
    cores1_m_i=random_MASs(1,nb_pts1_i,k1,d)[0]
    cores2_m_i=random_MASs(1,nb_pts2_i,k2,d)[0]
    FA1_i=np.random.rand(nb_pts1_i,k1)
    FA2_i=np.random.rand(nb_pts2_i,k2)
    w1_i=np.random.rand(nb_pts1_i,k1)
    w2_i=np.random.rand(nb_pts2_i,k2)
    w1_i/=w1_i.sum(-1)[:,None]
    w2_i/=w2_i.sum(-1)[:,None]

    
    cores1_m+=[cores1_m_i[0]]
    FA1+=[FA1_i]
    w1+=[w1_i]
    labels_svm1+=[np.random.choice(np.arange(1,41), size=nb_pts1_i, replace=False)]
    
    cores2_m+=[cores2_m_i[0]]
    FA2+=[FA2_i]
    w2+=[w2_i]
    labels_svm2+=[np.random.choice(np.arange(1,41), size=nb_pts2_i, replace=False)]
```

```python
C,C_nb_MAP=distances_FAS_svm(cores1_m,cores2_m,w1,w2,FA1,FA2,labels_svm1,labels_svm2,alpha=0.5,sum_or_max='sum')
```

```python
cost,sigma,C,C_nb_pts=distance_coresFA_svm(cores1_m,cores2_m,w1,w2,FA1,FA2,labels_svm1,labels_svm2,alpha=1,sum_or_max='sum')
print(cost)
```
